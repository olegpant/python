2.Payment x Total. Чи повʼязаний спосіб оплати з розміром чека.

3.Member vs Normal. Порівняти, скільки виручки та gross income приносять Member і Normal, але обов’язково поруч подивитися їхню кількість. Cередній чек, середню кількість товарів у чеку та улюблені категорії. Це покаже, чи Member дійсно цінніші для бізнесу, а не просто становлять більшу частку клієнтів.

```python
# Gender × Product line.
df.groupby(['Gender','Product line']).agg(
    count_invoice = ('Invoice ID', 'count'),
    avg_total_productLine = ('Total', 'mean'),
    total_quantity = ('Quantity', 'sum')
)
```

Перше, що кидається в очі, — Fashion accessories. У жінок кількість чеків вища: 96 проти 82 у чоловіків. Середній чек також вищий — близько 317 проти 291, а загальна кількість придбаних товарів становить 530 проти 372. Це свідчить про те, що жінки в цій категорії не тільки купують частіше, але й у середньому витрачають більше за одну покупку. Схожа ситуація спостерігається в категорії «Food & Beverage». У жінок середній чек значно вищий — близько 368 проти 273 у чоловіків, а загальна кількість придбаних товарів становить 514 проти 438. Кількість чеків при цьому порівнянна. Це свідчить про те, що жінки в цій категорії роблять більш великі покупки.Зворотна тенденція спостерігається в категорії Health & Beauty. Тут чоловіки активніші: кількість чеків становить 88 проти 64 у жінок, середній чек — близько 348 проти 290, а кількість придбаних товарів — 511 проти 343. Тобто в цій категорії чоловіки купують частіше і витрачають більше.

Для Fashion Accessories можна запропонувати посилити чоловічий сегмент: розширити асортимент товарів для чоловіків, протестувати окремі промо-викладки та акції, орієнтовані на чоловічу аудиторію. Водночас жіночий сегмент уже показує вищу активність за кількістю покупок, тому для нього можна використовувати таргетовані акції, набори товарів або крос-продажі, щоб додатково збільшити середній чек.
Для Food & Beverage доцільно використати високий потенціал жіночої аудиторії через персоналізовані акції, комплекти товарів і пропозиції на категорії, які вони купують найчастіше.
Для Health & Beauty варто окремо проаналізувати, які саме товари краще купують чоловіки, і перевірити, чи достатньо представлений асортимент для жінок. Якщо виявиться, що жіночий сегмент недоотримує потрібні продукти, можна розширити асортимент, додати цільові промоакції та покращити викладку товарів для збільшення продажів у цій категорії.

```python
# 2.Payment x Total
df.groupby(['Payment']).agg(
    count_invoice_by_payment = ('Invoice ID', 'count'),
    sum_total = ('Total','sum'),
    avg_total_payment = ('Total', 'mean'))
```

Згідно з таблицею, кількість інвойсів за різними способами оплати є приблизно однаковою, понад 300 для кожної категорії. Середній чек також майже не відрізняється й становить близько 300 незалежно від способу оплати. Отже, спосіб оплати не має помітного впливу на розмір середнього чека. Значної залежності між payment method і сумою покупки в цих даних не спостерігається.
Cпосіб оплати не є фактором, який помітно впливає на середній чек. Тому немає сенсу інвестувати в просування конкретного payment method саме з метою збільшення суми покупки. Для цього краще використовувати інші важелі: промоакції, крос-продажі, роботу з асортиментом і продуктовими категоріями.
Водночас усі способи оплати використовуються приблизно однаково часто. Тому бізнесу важливо підтримувати всі доступні канали оплати: картки, готівку та електронні гаманці. Щоб не створювати незручностей для значної частини покупців і не втрачати потенційні продажі.

```python
# 3.Member vs Norma
df.groupby(['Customer type']).agg(
    count_cutomerType = ('Customer type', 'count'),
    sum_total_by_cutomerType = ('Total', 'sum'),
    sum_gross_income = ('gross income', 'sum'),
    avg_quantity = ('Quantity', 'mean'),
    avg_total_by_cutomerType = ('Total', 'mean')

)
```

Кількість Member і звичайних клієнтів майже однакова: 501 проти 499. При цьому Member приносять трохи більше виручки, переважно завдяки вищому середньому чеку: 327 проти 318, а не через більшу кількість клієнтів.
Отже, Member є дещо ціннішим сегментом, але різниця поки невелика. Для бізнесу варто перевірити, чи можна посилити ефект програми лояльності через персональні пропозиції, бонуси, спеціальні акції та крос-продажі, щоб збільшити середній чек і частоту повторних покупок.
Водночас важливо не робити надто сильний висновок лише з цієї різниці: спочатку варто додатково порівняти частоту покупок, структуру категорій і витрати на одного клієнта.
