## Завдання 1: Порівняння Pandas vs Matplotlib

**Завдання:**
Побудуйте лінійний графік середньої кількості оренд помісячно впродовж всього періоду в даних двома способами:
1. Використовуючи Pandas (DataFrame.plot())
2. Використовуючи Matplotlib безпосередньо

В обох методах додайте маркери-кружечки. Можна також задати свій відмінний від стандартного колір.

Підказка: отримати потрібний формат даних найзручніше з методом датафрейму `resample`.

**Опишіть свої спостереження:** чим відрізняються 2 побудованих графіки? Який вам більше подобається?

```python
medium = df.resample('ME')['count'].mean()
medium.plot(
  figsize=(12,6),
  kind='line',
  title='Лінійний графік середньої кількості оренд',
  marker='o'
)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(medium.index,medium.values,marker='o')
plt.title('Лінійний графік середньої кількості оренд')
plt.show()
```

Обидва графіки побудовані на однакових даних, тому форма лінії ідентична. Проте є відмінності у відображенні та в коді.
У відображенні:
Підписи осі X (дати). Pandas розпізнає формат дати й підписує вісь зручно - назвами місяців (Jan, Apr, Jul) з роком знизу. Matplotlib показує дати «сирими» - цифрами (2011-01, 2011-04). Pandas читається легше.
Відступи від країв. У Pandas лінія починається майже від осі Y, у Matplotlib є невеликий відступ від краю графіка.
Мені більше подобається Matplotlib. Хоча коду доводиться писати трохи більше, підписи дат виглядають чіткіше й точніше (2011-01, 2012-04) — це дає ясніше уявлення про конкретний період, ніж скорочені назви місяців у Pandas. До того ж Matplotlib дає більше контролю над деталями графіка.

## Завдання 3: Subplot - 2x2 сітка графіків

**Завдання:**
Створіть сітку 2x2 з чотирма різними графіками, використовуючи `plt.subplot()`:
1. Лінійний графік середньої температури помісячно.
2. Стовпчикова діаграма середньої годинної кількості оренд за кварталами.
3. Гістограма вологості за всіма погодинними вимірами.
4. Scatter plot температури vs кількості оренд.

Кожен підграфік має містити всі необхідні підписи. Дашборд має містити назву.

```python
monthly_temp = df.resample('ME')['temp'].mean()
average_hour_count = df.groupby('season')['count'].mean()
average_hour_count = average_hour_count.rename({1:'Q1', 2:'Q2', 3:'Q3', 4:'Q4'})

plt.figure(figsize=(14,12))

plt.subplot(2,2,1)
plt.plot(monthly_temp.index, monthly_temp.values)
plt.title('Лінійний графік середньої температури помісячно')
plt.xticks(rotation=45)
plt.xlabel('Mісяць')
plt.ylabel('Середня температура')


plt.subplot(2,2,2)
average_hour_count.plot(kind='bar')
plt.title('Стовпчикова діаграма середньої годинної кількості оренд за кварталами.')
plt.xticks(rotation=0)
plt.xlabel('Квартал')
plt.ylabel('Кількість оренд')

plt.subplot(2,2,3)
plt.hist(df['humidity'],bins=20)
plt.title('Гістограма вологості за всіма погодинними вимірами')
plt.xlabel('Вологість')
plt.ylabel('Кількість вимірів')

plt.subplot(2,2,4)
plt.scatter(df['temp'],df['count'],alpha=0.6)
plt.title('Scatter plot температури vs кількості оренд')
plt.xlabel('Температура')
plt.ylabel('Кількість оренд')



plt.suptitle('Subplot - 2x2 сітка графіків')
plt.tight_layout()
plt.show()
```

## Завдання 4: Subplots - об'єктно-орієнтований підхід

**Завдання:**
Створіть той самий набір графіків, але використовуючи `fig, ax = plt.subplots()`.

**Дайте відповідь на питання своїми словами**
- Чим відрізняється підхід побудови кількох графіків на одній фігурі з `plt.subplots()` від `plt.subplot()`?

```python
fig, ax = plt.subplots(2,2, figsize=(14,12))

ax[0,0].plot(monthly_temp.index, monthly_temp.values)
ax[0,0].set_title('Лінійний графік середньої температури помісячно')
ax[0,0].tick_params(axis='x', rotation=45)
ax[0,0].set_xlabel('Mісяць')
ax[0,0].set_ylabel('Середня температура')

average_hour_count.plot(kind='bar', ax=ax[0,1])
ax[0,1].set_title('Стовпчикова діаграма середньої годинної кількості оренд за кварталами.')
ax[0,1].tick_params(axis='x', rotation=0)
ax[0,1].set_xlabel('Квартал')
ax[0,1].set_ylabel('Кількість оренд')

ax[1,0].hist(df['humidity'], bins=20)
ax[1,0].set_title('Гістограма вологості за всіма погодинними вимірами')
ax[1,0].set_xlabel('Вологість')
ax[1,0].set_ylabel('Кількість вимірів')

ax[1,1].scatter(df['temp'],df['count'],alpha=0.6)
ax[1,1].set_title('Scatter plot температури vs кількості оренд')
ax[1,1].set_xlabel('Температура')
ax[1,1].set_ylabel('Кількість оренд')

fig.suptitle('Subplots - обєктно-орієнтований підхід')
plt.tight_layout()
plt.show()
```
